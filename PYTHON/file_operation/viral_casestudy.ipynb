```python
# Find the most commonly used hashtag.

hashtag_list = [posts.get("Hashtag") for posts in viral_posts]
hashtag_count = {hash:hashtag_list.count(hash) for hash in hashtag_list}

import matplotlib.pyplot as plt
plt.figure(figsize=(12,8))
x = hashtag_count.keys()
y = hashtag_count.values()
plt.bar(x,y)
```
