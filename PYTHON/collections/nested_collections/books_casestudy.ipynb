```python
books = [
    ["The Hobbit", "Fantasy", 15, 1937, 4.8],           # 0
    ["Dune", "Sci-Fi", 18, 1965, 4.7],                 # 1
    ["Project Hail Mary", "Sci-Fi", 22, 2021, 4.9],    # 2
    ["Atomic Habits", "Self-Help", 16, 2018, 4.8],     # 3
    ["1984", "Dystopian", 12, 1949, 4.6],             # 4
    ["The Silent Patient", "Mystery", 14, 2019, 4.5],  # 5
    ["Sapiens", "History", 20, 2014, 4.7],            # 6
    ["Harry Potter", "Fantasy", 25, 1997, 4.9],        # 7
    ["Educated", "Memoir", 17, 2018, 4.4],             # 8
    ["Klara and the Sun", "Sci-Fi", 13, 2021, 4.2]     # 9
]
```

```python
# Print the titles of all books that contain more than 10 characters.

titles_mt_ten = [b[0] for b in books if len(b[0]) > 10]
titles_mt_ten
```

```python
# Find and print all titles that start with the word "The".

titles_sw_the = [b[0] for b in books if b[0].startswith("The")]
titles_sw_the
```

```python
# Create a set or list containing all unique genres present in the dataset (without duplicates).

unique_genres = {b[1] for b in books}
unique_genres
```

```python
# Count how many total books belong to the "Sci-Fi" genre.

count_of_sci = 0

for b in books:

    if b[1] == "Sci-Fi":
        count_of_sci +=1

print(count_of_sci)
```

```python
# Calculate the total cost if a customer buys one copy of every book in the dataset.

sum = 0

for b in books:

    sum += b[2]

print(sum)
```

```python
# Find and print the title and price of the cheapest book (lowest price).

price_list = [b[2] for b in books]
lowest_price = min(price_list)

# for b in books:

#     if b[2] == lowest_price:
#         print(f"title = {b[0]}")
#         print(f"price = {b[2]}")

title_lowest_price = [(b[:1],b[2]) for b in books if b[2] == lowest_price]
title_lowest_price
```

```python
# Print each book's title along with its age in years (using 2026 as the current year).

title_age = {b[0]:2026-b[3] for b in books}
title_age
```

```python
# Filter and display all books published before the year 2000.

books_before_2000 = [b[0] for b in books if b[3] < 2000]
books_before_2000
```

```python
# Calculate the average rating of all 10 books in the dataset.

sum_of_rating = 0
count_of_rating = 0

for b in books:

    sum_of_rating+=b[4]
    count_of_rating+=1

avg_of_rating = sum_of_rating/count_of_rating
avg_of_rating
```

```python
# Count how many books have a rating of 4.8 or higher.

count_of_higher = 0

for b in books:

    if b[4] >= 4.8:
        count_of_higher+=1

count_of_higher
```

```python
# Write a script to find the title of the most expensive "Sci-Fi" book published after 2010

books_after_2010 = [b for b in books if b[1] == "Sci-Fi" and b[-2] > 2010]

most_expensive_2010 = books_after_2010[0][2]

title_expensive = [b[0] for b in books_after_2010 if b[2] >= most_expensive_2010]
title_expensive
```
