```python
movies = [
    ["dangal", "hindi", 161, 2016, 8.3],       # 0
    ["premam", "malayalam", 156, 2015, 8.3],   # 1
    ["inception", "english", 148, 2010, 8.8],  # 2
    ["vikram", "tamil", 175, 2022, 8.3],       # 3
    ["parasite", "korean", 132, 2019, 8.5],    # 4
    ["rrr", "telugu", 187, 2022, 7.8],         # 5
    ["drishyam", "malayalam", 160, 2013, 8.3], # 6
    ["avatar", "english", 162, 2009, 7.9],     # 7
    ["kumbalangi nights", "malayalam", 135, 2019, 8.5], # 8
    ["sholay", "hindi", 204, 1975, 8.1]        # 9
]
```

```python
# Find by Language: Print the titles of all Malayalam movies.

malayalam_titles = [m[0] for m in movies if m[1] == "malayalam"]
malayalam_titles
```

```python
# Find by Year: Print the titles of all movies released after the year 2015.

movies_after_2015 = [m[0] for m in movies if m[-2] > 2015]
movies_after_2015
```

```python
# Find by Rating: Count how many movies have a rating of 8.5 or higher.

count = 0

for m in movies:

    if m[-1] >= 8.5:
        count+=1

print(f"count of movies have a rating of 8.5 or higher = {count}")
```

```python
# Average Duration: Calculate the average duration (in minutes) of all the movies in the dataset.

sum = 0
count_of_duration = 0

for m in movies:

    sum+=m[2]
    count_of_duration+=1

avg = sum/count_of_duration
avg
```

```python
# Multiple Conditions: Print the titles of movies that are longer than 150 minutes AND have a rating greater than 8.0.

titles_with_both = [m[0] for m in movies if m[2] > 150 and m[-1] > 8.0]
titles_with_both
```

```python
# Highest Rating: Find and print the entire record for the movie with the highest rating.

highest_rating_list = [m[-1] for m in movies]
highest_rating = max(highest_rating_list)

record_highest = [m for m in movies if m[-1] == highest_rating]
record_highest
```
