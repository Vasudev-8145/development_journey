```python
# display all owners

all_owners = {e.get("owner") for e in expenses}
all_owners
```

```python
# all transactions

transactions = {e.get("paymentmethod") for e in expenses}
transactions
```

```python
# display details of owner name startswith "a"

name_filter = [e for e in expenses if e.get("owner").startswith("A")]
name_filter
```

```python
# owner-wise count

all_owners_list = [e.get("owner") for e in expenses]

owner_count = {o:all_owners_list.count(o) for o in all_owners_list}
owner_count
```

```python
import matplotlib.pyplot as plt

x = owner_count.keys()
y = owner_count.values()

plt.bar(x,y)
```

```python
# owner-wise transaction sum

owner_summary = {}

for e in expenses:

    owner = e.get("owner")
    amount = e.get("amount")

    if owner in owner_summary:
        owner_summary[owner] += amount

    else:
        owner_summary[owner] = amount

owner_summary
```

```python
x = owner_summary.keys()
y = owner_summary.values()

plt.bar(x,y)
```

```python
# highest transaction owner

transactions = {e.get("amount") for e in expenses}
highest_transaction = max(transactions)

highest_transaction_owner = [e for e in expenses if e["amount"] == highest_transaction]
highest_transaction_owner
```

```python
# transaction with lowest amount

lowest_transaction = min([e.get("amount") for e in expenses])

lowest_transaction_owner = [e for e in expenses if e["amount"] == lowest_transaction]
lowest_transaction_owner
```

```python
# average transaction

sum = 0
count = 0

for e in expenses:

    sum += e.get("amount")
    count += 1

avg_transaction = sum/count
avg_transaction
```
