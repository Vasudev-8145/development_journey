### movie casestudy

```python
movies = [

        ["kgf","kannada",150,2020,8],
        ["dhrishyam 3","malayalam",170,"2026","8"],
        ["spiderman","english",160,"2026","9"],
        ["vikram","tamil",175,"2022","8"],
        ["Vazha 2","malayalam",150,"2025","10"],

        ]

```

```python
# all movie titles

all_titles = [m[0] for m in movies]
all_titles
```

```python
# all movie languages

all_languages = {m[1] for m in movies}
all_languages
```

```python
# all movie titles in uppercase

titles = [m[0].upper() for m in movies]
titles
```

```python
# movie titles startswith v

titles = [m[0] for m in movies if m[0].lower().startswith("v")]
titles
```

```python
# display malayalam movies

malayalam_movies = [m[0] for m in movies if m[1] == "malayalam"]
malayalam_movies
```

```python
# language-wise count

languages = [m[1] for m in movies]

lc = {l:languages.count(l) for l in languages}
lc
```

```python
import matplotlib.pyplot as plt

x = lc.keys()
y = lc.values()

plt.bar(x,y)
```

```python
# highest run time movie

movies = [

        ["kgf","kannada",150,2020,8],
        ["dhrishyam 3","malayalam",170,"2026","8"],
        ["spiderman","english",160,"2026","9"],
        ["vikram","tamil",175,"2022","8"],
        ["Vazha 2","malayalam",150,"2025","10"],

        ]


runtime_list = [m[2] for m in movies]
highest_runtime = max(runtime_list)
highest_runtime_movie = {m[0]:m[2] for m in movies if m[2] == highest_runtime}
highest_runtime_movie

```

```python
# lowest runtime movie

lowest_runtime = min(runtime_list)
lowest_runtime_movie = {m[0]:m[2] for m in movies if m[2] == lowest_runtime}
lowest_runtime_movie
```

```python
# average runtime

count = 0
sum = 0

for m in movies:

    sum+=m[2]
    count+=1

average = sum/count
print(f"average runtime = {average}")
```

```python
# movies above average runtime

above_avrg = {m[0] for m in movies if m[2]>average}
above_avrg
```

```python
# movies below average runtime

below_avrg = {m[0] for m in movies if m[2]<average}
below_avrg
```
